# Plotting

A code cell whose output is an embedded PNG, produced by Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 4 * np.pi, 400)

fig, ax = plt.subplots(figsize=(6, 3.5))
ax.plot(x, np.sin(x), label="sin(x)")
ax.plot(x, np.sin(x) * np.exp(-x / 10), label="damped")
ax.set_xlabel("x")
ax.set_ylabel("amplitude")
ax.set_title("A sine wave, and a damped sine wave")
ax.legend()
ax.grid(alpha=0.3)
fig.tight_layout()
```
