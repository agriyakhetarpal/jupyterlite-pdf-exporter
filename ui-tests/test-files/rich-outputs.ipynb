# Rich outputs

This test notebook has outputs as saved by pandas, xarray, folium, Plotly, and ipywidgets, so that we can check what the exporter does with HTML and with outputs that only work in a browser. The HTML was trimmed by hand to keep the file small, but the structure is what each library writes.

```python
import pandas as pd

df = pd.DataFrame({'city': ['Paris', 'Delhi'], 'temp': [20.6, 19.3]})
df
```

```python
import numpy as np
import xarray as xr

da = xr.DataArray(np.zeros(3), dims='lat')
da
```

```python
import folium

m = folium.Map(location=[48.85, 2.35], zoom_start=12)
m
```

```python
import plotly.express as px

fig = px.scatter(df, x='city', y='temp')
fig
```
